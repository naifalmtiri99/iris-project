## Task 6: Export Analysis Results

```python
import os
os.makedirs("output", exist_ok=True)
```

```python
species_summary = df.groupby("species").mean(numeric_only=True)

species_summary.to_csv("output/species_summary.csv")

species_summary
```

```python
filtered_iris = df[df["sepal_length"] > 6]

filtered_iris.to_csv("output/filtered_iris.csv", index=False)

filtered_iris
```

```python
import os

print(os.path.exists("output/species_summary.csv"))
print(os.path.exists("output/filtered_iris.csv"))
```

## Task 6: Export Analysis Results

The `output` folder was created to store the analysis results.

The mean measurements for each Iris species were calculated using `groupby()` and saved to `output/species_summary.csv`.

All flowers with a `sepal_length` greater than 6 were filtered and saved to `output/filtered_iris.csv`.

Finally, both files were checked, and the results confirmed that they were created successfully.

## Bar Chart: Average Petal Length by Iris Species

```python
import matplotlib.pyplot as plt

species_summary["petal_length"].plot(kind="bar")

plt.title("Average Petal Length by Iris Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

The bar chart compares the average petal length of the three Iris species. Virginica has the highest average petal length, followed by versicolor, while setosa has the lowest.
